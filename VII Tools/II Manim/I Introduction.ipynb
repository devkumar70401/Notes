# Introduction to Manim
Run the cells below to render animations directly inside the notebook.

```python
# Step 1: Import Manim to register the %%manim cell magic
from manim import *
```

```python
%%manim -ql -v WARNING CreateCircle

class CreateCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.5)  # set the color and transparency
        self.play(Create(circle))  # show the circle on screen
```

```python
%%manim -ql -v Warning SquareToCircle

class SquareToCircle(Scene):
    def construct(self):
        square = Square()
        square.rotate(PI / 4)
        circle = Circle()
        circle.set_fill(color=PINK, opacity=0.3)

        self.play(Create(square))
        self.play(Transform(square, circle))
        self.play(FadeOut(square))
```
