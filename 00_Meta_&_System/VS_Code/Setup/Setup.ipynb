# Professional VS Code Workspace Setup (Multi-Repo + Multi-GitHub + Separate venv)

---

# 1. Create Main Workspace Folder

Create one permanent workspace folder.

```bash
mkdir -p ~/SE
```

Go inside it:

```bash
cd ~/SE
```

---

# 2. Create Project/Utility/Notes Folders

Example:

```bash
mkdir Notes
mkdir trekking-management-app-v1
mkdir DL_Proj
mkdir MAD-1_Proj
mkdir MAD-2_Proj
mkdir MLP_Proj
```

You can create/remove folders anytime later.

---

# 3. Open Workspace in VS Code

```bash
code ~/SE
```

IMPORTANT:
- `SE` is ONLY a workspace folder
- `SE` should NOT be a git repository

NEVER RUN:

```bash
git init
```

inside:

```bash
~/SE
```

---

# 4. Save VS Code Workspace

Inside VS Code:

```text
File → Save Workspace As
```

Save as:

```text
SE.code-workspace
```

inside:

```text
~/SE
```

Final structure:

```text
~/SE
│
├── SE.code-workspace
├── Notes/
├── trekking-management-app-v1/
├── DL_Proj/
├── MAD-1_Proj/
├── MAD-2_Proj/
└── MLP_Proj/
```

# 6. Setup Each Project Folder Independently

Each project folder should be its OWN:

- git repository
- virtual environment
- dependency environment

Example:

```text
trekking-management-app-v1/

├── .git/
├── .venv/
├── requirements.txt
├── .gitignore
├── README.md
└── src/
```

# 7. Initialize Git Repo

Go inside project folder:

```bash
cd ~/SE/trekking-management-app-v1
```

```bash
git init
```

---

# 8. Create Separate Virtual Environment

Create venv:

```bash
python3 -m venv .venv
```

Activate:

```bash
source .venv/bin/activate
```


IMPORTANT:
- Every project should have separate `.venv`
- Never share venv between projects

- Enter
- Ctrl + X

# 14. Multi-GitHub Account Setup (Professional SSH Setup)

Recommended when using:
- personal GitHub account
- academic/company GitHub account

Example:

| Purpose | Account |
|---|---|
| Notes / Personal Projects | Personal GitHub |
| Academic Projects | Academic GitHub |

---

# 15. Create Separate SSH Keys

## Personal GitHub

```bash
ssh-keygen -t ed25519 -C "personal_github"
```

Save as:

```text
~/.ssh/github_personal
```

---

## Academic GitHub

```bash
ssh-keygen -t ed25519 -C "academic_github"
```

Save as:

```text
~/.ssh/github_academic
```

---

# 16. Start SSH Agent

```bash
eval "$(ssh-agent -s)"
```

---

# 17. Add SSH Keys

```bash
ssh-add ~/.ssh/github_personal
```

```bash
ssh-add ~/.ssh/github_academic
```

---

# 18. Configure SSH

Create config:

```bash
nano ~/.ssh/config
```

Paste:

```config
# Personal GitHub
Host github-personal
    HostName github.com
    User git
    IdentityFile ~/.ssh/github_personal

# Academic GitHub
Host github-academic
    HostName github.com
    User git
    IdentityFile ~/.ssh/github_academic
```

Save:
- Ctrl + O
- Enter
- Ctrl + X

---

# 19. Add SSH Public Keys To GitHub

## Personal Account

Copy key:

```bash
cat ~/.ssh/github_personal.pub
```

Add to:

https://github.com/settings/keys

while logged into PERSONAL GitHub account.

---

## Academic Account

Copy key:

```bash
cat ~/.ssh/github_academic.pub
```

Add to:

https://github.com/settings/keys

while logged into ACADEMIC GitHub account.
