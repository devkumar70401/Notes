# Introduction

## Running Python

- Python programs run inside an interpreter.
- The interpreter is a simple "console-based" application that normally starts from a command shell

## Interactive Mode

- The interpreter runs a "read-eval" loop

```python
# REPL 
while True:
    user_input = read_line(">>> ")  # Read
    result = evaluate(user_input)   # Eval
    print(result)                   # Print
    # Loop repeats until you type exit() or Ctrl + D 
```

````python
print('hello world')
````

````python
for i in range(5):
    print(i)
````

- Executes simple statements typed in directly
- Very useful for debugging, exploration

## Creating programs

- Programs are put in .py files

```python
# helloworld.py 
print('hello world') 
```

- Source files are simple text files


## Running Programs

- Command line

```bash
python3 helloworld.py 
hello world
```

## python -i

- For debugging, use python -i  
```bash
python3 -i helloworld.py 
```

- Runs your program and then enters the python interactive shell afterwards
- quite useful for debugging, testing, etc.
