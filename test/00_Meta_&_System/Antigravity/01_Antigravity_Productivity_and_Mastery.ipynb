## 📌 1. Core Architecture & Workflow Capabilities

Antigravity is not a passive chatbot—it is an **active Linux agent** embedded inside your workspace (`/home/dev/SE`).

### Key Powers:
1. **Autonomous File Management**: Creates `.ipynb`, `.md`, `.py` scripts directly in your notes and project folders.
2. **Terminal & Environment Control**: Installs packages (`pip`, `apt`), manages virtualenvs (`.venv`), runs tests (`pytest`), and executes Docker containers.
3. **Browserless GitHub Integration**: Uses `gh` CLI and SSH keypairs to create, pull, push, and list GitHub repositories without web browser prompts.
4. **Subagents & Background Tasks**: Spawns background research subagents to search codebases or docs while you keep working.


## 🛠️ 2. Essential Slash Commands for Accelerated Learning

- **/plan**: Generates a high-level technical plan before starting complex ML projects.
- **/goal**: Runs an autonomous session where Antigravity works continuously until a task is 100% finished.
- **/grill-me**: Initiates an active-recall interview session to test your knowledge on ML theory, math, or code.
- **/schedule**: Sets recurring timer notifications or background tasks.
- **/learn**: Persists custom coding styles, mentor personas, or codebase rules into `.agents/rules/`.
